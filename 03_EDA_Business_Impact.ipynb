```python
import pandas as pd

excel_file_path = "reference_data/vulnerability_contextualization_scanner_sample_updated.xlsx"

# Load the Excel file into the pandas DataFrame variable
df = pd.read_excel(excel_file_path, engine="openpyxl")

# Execute info invocation to inspect columns, non-null counts, and memory footprints
df.info()
```

```python
df.nunique()
```

```python
df['Contextual priority'].value_counts()
```

```python
counts = df['Exposure classification'].value_counts()

ax = counts.plot(
    kind='pie',
    autopct=lambda pct: f'{int(round(pct * counts.sum() / 100))}',
    wedgeprops={'width': 0.4},
    startangle=90,
    ylabel='',
    title='Count by Exposure Classification'
)

ax.set_aspect('equal')
ax.figure.tight_layout()
```

```python
df[['Critical assets']]
```

```python
critical_by_group = (
    df.groupby('Asset group')['Critical assets']
      .sum()
      .sort_values(ascending=False)
)

ax = critical_by_group.plot(kind='bar', title='Critical Assets by Asset Group')
ax.set_xlabel('Asset Group')
ax.set_ylabel('Critical Assets')
ax.figure.tight_layout()
```
